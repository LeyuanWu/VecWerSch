### 计算公式

The exterior gravitational potential, along with its vectors and gradient tensors, of a polyhedron can be forward calculated using the algorithm proposed by Werner and Scheeres (1996). The algorithm is derived in closed form so that there is no error produced during the process of forward calculating theoretically. Assuming the distribution of density inside a polyhedron, expressed as 
in the following equations, is constant, all facets of the polyhedron are triangles and vertices of a facet are ordered anticlockwise as one observes the facet from outside of the polyhedron. Then the polyhedron’s exterior gravitational potential V, gradient and gradient tensors are given as below, in which G is the gravitational constant. Gradient directions here are x, y and z in the Cartesian coordinate system.

$$\begin{aligned}&V=\frac{1}{2}{G}\rho \left( {\sum \nolimits _{e=1}^{n}} \vec {r}_{e}\cdot {\mathbf{E}}_{e} \cdot \vec {r}_{e} \cdot {L}_{e}\right. \nonumber \\&\qquad \left. +{\sum \nolimits _{f=1}^{m}} \vec {r}_{f}\cdot {\mathbf{F}}_{f}\cdot \vec {{r}_{f} }\cdot {\omega }_{f} \right) , \end{aligned}$$

$$\begin{aligned}&\nabla V = -G \rho \left( {\sum \nolimits _{e=1}^{n}{\mathbf{E}}_{e}\cdot \vec {{r}_{e} }\cdot L_{e} +\sum \nolimits _{{f=1}}^{m} {\mathbf{F}}_{f} \cdot \vec {r}_{f} \cdot {\omega }_{f} } \right) , \nonumber \\ \end{aligned}$$

$$\begin{aligned}&\nabla \nabla V = -G\rho \left( \sum \nolimits _{e=1}^{n} {\mathbf{E}}_{e} \cdot {L}_{e} +\sum \nolimits _{f=1}^{m} {\mathbf{F}}_{f} \cdot {\omega }_{f} \right) . \end{aligned}$$
