**Appendix**

*Green's third identity*

$$
V(\mathbf{r}) = \frac{1}{4\pi} \oint_S 
\left[ 
V(\mathbf{r}') \frac{\partial}{\partial n'} \left( \frac{1}{|\mathbf{r} - \mathbf{r}'|} \right)
    - \frac{1}{|\mathbf{r} - \mathbf{r}'|} \frac{\partial V(\mathbf{r}')}{\partial n'}
\right] dS' \tag{A1}
$$
